```python
from features import PEFeatureExtractor
import lightgbm as lgb
import numpy as np
```

```python
extractor = PEFeatureExtractor(feature_version=2)

with open("D:/OktaVerifySetup-6.0.0.0-474cbc4.exe", "rb") as f:
    bytez = f.read()

features = extractor.feature_vector(bytez)
print("Extracted feature shape:", features.shape)
```

```python
model = lgb.Booster(model_file="model/computer_virus_detection_ember_lightGBM.txt")

prediction = model.predict(np.array([features]))[0]

print("\n====== RESULT ======")
print("Malware probability:", prediction)
print("Malware verdict:", "MALWARE" if prediction >= 0.5 else "BENIGN")
print("====================")
```
